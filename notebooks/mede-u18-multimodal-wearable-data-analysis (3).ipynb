## CP1 – Temporal alignment and coverage report

The two datasets are aligned independently because they contain different participants.
The common analysis unit is person × temporal window.

For CP1, the goal is to align motion and cardiac information in time,
without extracting the final feature matrix yet.

```python
# CP1 - PAMAP2 temporal alignment

WINDOW_SECONDS = 5

pamap2_aligned = pamap2.copy()

# Activity 0 = transient / undefined activity -> exclude
pamap2_aligned = pamap2_aligned[
    pamap2_aligned["activity_id"] != 0
].copy()

# Create a 5-second temporal grid for each subject
pamap2_aligned["window_id"] = (
    pamap2_aligned["timestamp"] // WINDOW_SECONDS
).astype(int)

# Interpolate heart rate independently for each subject
pamap2_aligned["heart_rate_aligned"] = (
    pamap2_aligned
    .groupby("subject_id")["heart_rate"]
    .transform(lambda x: x.interpolate(limit_direction="both"))
)

print("PAMAP2 rows after removing activity 0:", len(pamap2_aligned))

print(
    "Missing HR before alignment:",
    f"{pamap2_aligned['heart_rate'].isna().mean() * 100:.2f}%"
)

print(
    "Missing HR after alignment:",
    f"{pamap2_aligned['heart_rate_aligned'].isna().mean() * 100:.2f}%"
)

pamap2_aligned[
    [
        "subject_id",
        "timestamp",
        "window_id",
        "activity_id",
        "heart_rate",
        "heart_rate_aligned"
    ]
].head(15)
```

```python
# CP1 - PAMAP2 aligned window table

pamap2_windows = (
    pamap2_aligned
    .groupby(["subject_id", "window_id"])
    .agg(
        start_time=("timestamp", "min"),
        end_time=("timestamp", "max"),
        activity_id=("activity_id", lambda x: x.mode().iloc[0]),
        imu_samples=("timestamp", "size"),
        hr_raw_samples=("heart_rate", "count"),
        heart_rate=("heart_rate_aligned", "mean")
    )
    .reset_index()
)

# Coverage: a valid aligned window contains motion samples
# and cardiac information after temporal alignment
pamap2_windows["has_motion"] = pamap2_windows["imu_samples"] > 0
pamap2_windows["has_cardiac"] = pamap2_windows["heart_rate"].notna()
pamap2_windows["complete_multimodal"] = (
    pamap2_windows["has_motion"] &
    pamap2_windows["has_cardiac"]
)

print("PAMAP2 aligned table")
print("Unit: subject × 5-second window")
print("Total windows:", len(pamap2_windows))
print(
    "Complete multimodal windows:",
    pamap2_windows["complete_multimodal"].sum()
)
print(
    "Coverage:",
    f"{pamap2_windows['complete_multimodal'].mean() * 100:.2f}%"
)

pamap2_windows.head(10)
```

```python
# CP1 - MHEALTH temporal alignment

WINDOW_SECONDS = 5
MHEALTH_HZ = 50
SAMPLES_PER_WINDOW = WINDOW_SECONDS * MHEALTH_HZ  # 250

mhealth_aligned = mhealth.copy()

# Activity 0 = null / transition periods -> exclude
mhealth_aligned = mhealth_aligned[
    mhealth_aligned["activity"] != 0
].copy()

# Rebuild sample position independently for each subject
mhealth_aligned["sample_id"] = (
    mhealth_aligned
    .groupby("subject_id")
    .cumcount()
)

# Temporal grid: 250 samples = 5 seconds at 50 Hz
mhealth_aligned["window_id"] = (
    mhealth_aligned["sample_id"] // SAMPLES_PER_WINDOW
).astype(int)

# Check availability of motion and ECG at sample level
motion_cols = ["chest_acc_x", "chest_acc_y", "chest_acc_z"]
ecg_cols = ["ecg_lead_1", "ecg_lead_2"]

mhealth_aligned["motion_valid"] = (
    mhealth_aligned[motion_cols].notna().all(axis=1)
)

mhealth_aligned["ecg_valid"] = (
    mhealth_aligned[ecg_cols].notna().all(axis=1)
)

print("MHEALTH rows after removing activity 0:", len(mhealth_aligned))
print(
    "Missing motion samples:",
    f"{(1 - mhealth_aligned['motion_valid'].mean()) * 100:.2f}%"
)
print(
    "Missing ECG samples:",
    f"{(1 - mhealth_aligned['ecg_valid'].mean()) * 100:.2f}%"
)
```

```python
# Build aligned table: one row = subject × 5-second window

mhealth_windows = (
    mhealth_aligned
    .groupby(["subject_id", "window_id"])
    .agg(
        activity=("activity", lambda x: x.mode().iloc[0]),
        samples=("sample_id", "size"),
        motion_samples=("motion_valid", "sum"),
        ecg_samples=("ecg_valid", "sum")
    )
    .reset_index()
)

# A full 5-second MHEALTH window contains 250 samples
mhealth_windows["full_window"] = (
    mhealth_windows["samples"] == SAMPLES_PER_WINDOW
)

mhealth_windows["complete_multimodal"] = (
    (mhealth_windows["motion_samples"] == mhealth_windows["samples"]) &
    (mhealth_windows["ecg_samples"] == mhealth_windows["samples"])
)

print("\nMHEALTH aligned table")
print("Unit: subject × 5-second window")
print("Total windows:", len(mhealth_windows))
print("Full 5-second windows:", mhealth_windows["full_window"].sum())
print(
    "Complete multimodal windows:",
    mhealth_windows["complete_multimodal"].sum()
)
print(
    "Multimodal coverage:",
    f"{mhealth_windows['complete_multimodal'].mean() * 100:.2f}%"
)

mhealth_windows.head(10)
```

```python
# CP1 - Final alignment coverage report

# PAMAP2: full 5-second window = 500 IMU samples
pamap2_windows["full_window"] = (
    pamap2_windows["imu_samples"] == 500
)

pamap2_valid = pamap2_windows[
    pamap2_windows["full_window"] &
    pamap2_windows["complete_multimodal"]
].copy()

# MHEALTH: full 5-second window = 250 samples
mhealth_valid = mhealth_windows[
    mhealth_windows["full_window"] &
    mhealth_windows["complete_multimodal"]
].copy()

coverage_report = pd.DataFrame({
    "Dataset": ["PAMAP2", "MHEALTH"],
    "Subjects": [
        pamap2_valid["subject_id"].nunique(),
        mhealth_valid["subject_id"].nunique()
    ],
    "Total windows": [
        len(pamap2_windows),
        len(mhealth_windows)
    ],
    "Full 5s windows": [
        pamap2_windows["full_window"].sum(),
        mhealth_windows["full_window"].sum()
    ],
    "Final aligned windows": [
        len(pamap2_valid),
        len(mhealth_valid)
    ],
    "Coverage (%)": [
        len(pamap2_valid) / len(pamap2_windows) * 100,
        len(mhealth_valid) / len(mhealth_windows) * 100
    ]
})

coverage_report["Coverage (%)"] = (
    coverage_report["Coverage (%)"].round(2)
)

coverage_report
```

```python
# CP2 - Verify data and alignment outputs from CP1

print("PAMAP2 original:", pamap2.shape)
print("MHEALTH original:", mhealth.shape)

print("\nPAMAP2 valid windows:", len(pamap2_valid))
print("MHEALTH valid windows:", len(mhealth_valid))

print("\nPAMAP2 columns:")
print(pamap2.columns.tolist())

print("\nMHEALTH columns:")
print(mhealth.columns.tolist())
```

```python
# CP2 - Correct MHEALTH temporal windows

WINDOW_SECONDS = 5
MHEALTH_HZ = 50
SAMPLES_PER_WINDOW = WINDOW_SECONDS * MHEALTH_HZ

mhealth_corrected = mhealth.copy()

# Preserve the original sample position for each subject
mhealth_corrected["sample_id"] = (
    mhealth_corrected.groupby("subject_id").cumcount()
)

# Construct windows before removing activity 0
mhealth_corrected["window_id"] = (
    mhealth_corrected["sample_id"] // SAMPLES_PER_WINDOW
)

# Remove unlabeled / transition samples
mhealth_corrected = mhealth_corrected[
    mhealth_corrected["activity"] != 0
].copy()

# Count samples in each window
mhealth_window_counts = (
    mhealth_corrected
    .groupby(["subject_id", "window_id"])
    .size()
    .reset_index(name="samples")
)

# Keep only full 5-second windows
mhealth_full_windows = mhealth_window_counts[
    mhealth_window_counts["samples"] == SAMPLES_PER_WINDOW
].copy()

print("Previous CP1 valid windows:", len(mhealth_valid))
print("Corrected full windows:", len(mhealth_full_windows))
print("Subjects retained:", mhealth_full_windows["subject_id"].nunique())

display(mhealth_full_windows.head(10))
```

```python

# CP2 - Validate PAMAP2 temporal windows

import numpy as np
import pandas as pd

pamap2_check = pamap2.copy()

WINDOW_SECONDS = 5
EXPECTED_SAMPLES = 500  # PAMAP2 IMU: 100 Hz

pamap2_check = pamap2_check[
    pamap2_check["activity_id"] != 0
].copy()

# Keep original timestamps when assigning windows
pamap2_check["window_id"] = (
    np.floor(pamap2_check["timestamp"] / WINDOW_SECONDS)
    .astype(int)
)

pamap2_window_check = (
    pamap2_check
    .groupby(["subject_id", "window_id"])
    .agg(
        samples=("timestamp", "size"),
        start_time=("timestamp", "min"),
        end_time=("timestamp", "max"),
        activities=("activity_id", "nunique")
    )
    .reset_index()
)

# A valid window must have all samples and one activity
pamap2_window_check["valid"] = (
    (pamap2_window_check["samples"] == EXPECTED_SAMPLES)
    & (pamap2_window_check["activities"] == 1)
)

pamap2_full_windows = pamap2_window_check[
    pamap2_window_check["valid"]
].copy()

print("Total candidate windows:", len(pamap2_window_check))
print("Valid 5-second windows:", len(pamap2_full_windows))
print("Subjects retained:", pamap2_full_windows["subject_id"].nunique())
print(
    "Coverage:",
    f"{len(pamap2_full_windows) / len(pamap2_window_check) * 100:.2f}%"
)

display(pamap2_full_windows.head(10))
```

```python

# CP2 - PAMAP2 feature extraction
# One row = subject x valid 5-second window

import numpy as np
import pandas as pd

WINDOW_SECONDS = 5

# Keep only samples belonging to validated windows
pamap2_features_data = pamap2.copy()
pamap2_features_data = pamap2_features_data[
    pamap2_features_data["activity_id"] != 0
].copy()

pamap2_features_data["window_id"] = np.floor(
    pamap2_features_data["timestamp"] / WINDOW_SECONDS
).astype(int)

valid_keys = pamap2_full_windows[
    ["subject_id", "window_id"]
].drop_duplicates()

pamap2_features_data = pamap2_features_data.merge(
    valid_keys,
    on=["subject_id", "window_id"],
    how="inner",
    validate="many_to_one"
)

# Chest accelerometer: x, y, z
acc_cols = [
    "chest_acc16_x",
    "chest_acc16_y",
    "chest_acc16_z"
]

# Acceleration magnitude
pamap2_features_data["acc_magnitude"] = np.sqrt(
    (pamap2_features_data[acc_cols] ** 2).sum(axis=1)
)

# Use observed HR values; no artificial interpolation
group_cols = ["subject_id", "window_id"]

pamap2_X = (
    pamap2_features_data
    .groupby(group_cols)
    .agg(
        activity=("activity_id", "first"),
        acc_mean=("acc_magnitude", "mean"),
        acc_std=("acc_magnitude", "std"),
        acc_min=("acc_magnitude", "min"),
        acc_max=("acc_magnitude", "max"),
        hr_mean=("heart_rate", "mean"),
        hr_std=("heart_rate", "std"),
        hr_samples=("heart_rate", "count")
    )
    .reset_index()
)

print("PAMAP2 feature matrix:", pamap2_X.shape)
print("Subjects:", pamap2_X["subject_id"].nunique())
print("Missing values:")
print(pamap2_X.isna().sum())

display(pamap2_X.head(10))
```

```python

# CP2 - MHEALTH feature extraction
# One row = subject x valid 5-second window

import numpy as np
import pandas as pd

WINDOW_SAMPLES = 250  # 5 seconds at 50 Hz

mhealth_features_data = mhealth.copy()

# Preserve original sample positions within each subject
mhealth_features_data["sample_id"] = (
    mhealth_features_data.groupby("subject_id").cumcount()
)

mhealth_features_data["window_id"] = (
    mhealth_features_data["sample_id"] // WINDOW_SAMPLES
)

# Retain only previously validated windows
valid_keys = mhealth_full_windows[
    ["subject_id", "window_id"]
].drop_duplicates()

mhealth_features_data = mhealth_features_data.merge(
    valid_keys,
    on=["subject_id", "window_id"],
    how="inner",
    validate="many_to_one"
)

# Chest acceleration magnitude
acc_cols = [
    "chest_acc_x",
    "chest_acc_y",
    "chest_acc_z"
]

mhealth_features_data["acc_magnitude"] = np.sqrt(
    (mhealth_features_data[acc_cols] ** 2).sum(axis=1)
)

# ECG signal amplitude statistics
# These are NOT heart-rate or HRV measurements
mhealth_X = (
    mhealth_features_data
    .groupby(["subject_id", "window_id"])
    .agg(
        activity=("activity", "first"),
        acc_mean=("acc_magnitude", "mean"),
        acc_std=("acc_magnitude", "std"),
        acc_min=("acc_magnitude", "min"),
        acc_max=("acc_magnitude", "max"),
        ecg_mean=("ecg_lead_1", "mean"),
        ecg_std=("ecg_lead_1", "std"),
        ecg_min=("ecg_lead_1", "min"),
        ecg_max=("ecg_lead_1", "max")
    )
    .reset_index()
)

print("MHEALTH feature matrix:", mhealth_X.shape)
print("Subjects:", mhealth_X["subject_id"].nunique())

print("Missing values:")
print(mhealth_X.isna().sum())

print("Samples per window:")
print(
    mhealth_features_data
    .groupby(["subject_id", "window_id"])
    .size()
    .value_counts()
    .head()
)

display(mhealth_X.head(10))
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

datasets = {
    "PAMAP2": (
        pamap2_X,
        ["acc_mean", "acc_std", "acc_min", "acc_max",
         "hr_mean", "hr_std"]
    ),
    "MHEALTH": (
        mhealth_X,
        ["acc_mean", "acc_std", "acc_min", "acc_max",
         "ecg_mean", "ecg_std", "ecg_min", "ecg_max"]
    )
}

pca_results = {}

for name, (df, features) in datasets.items():

    print(f"\n===== {name} =====")

    X = df[features].copy()
    X = X.replace([np.inf, -np.inf], np.nan)

    print("Observations:", len(X))
    print("Features:", len(features))
    print("Missing values:", X.isna().sum().sum())

    if X.isna().any().any():
        raise ValueError(
            f"{name}: missing values in PCA features"
        )

    # Standardization
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # PCA
    pca = PCA()
    X_pca = pca.fit_transform(X_scaled)

    explained = pca.explained_variance_ratio_
    cumulative = np.cumsum(explained)

    loadings = pd.DataFrame(
        pca.components_.T,
        index=features,
        columns=[
            f"PC{i+1}" for i in range(len(features))
        ]
    )

    pca_results[name] = {
        "scaler": scaler,
        "pca": pca,
        "transformed": X_pca,
        "loadings": loadings,
        "cumulative_variance": cumulative
    }

    print(
        "PC1 variance:",
        round(explained[0] * 100, 2), "%"
    )

    print(
        "PC2 variance:",
        round(explained[1] * 100, 2), "%"
    )

    print(
        "PC1 + PC2:",
        round(cumulative[1] * 100, 2), "%"
    )

    print(
        "Components for 90% variance:",
        np.searchsorted(cumulative, 0.90) + 1
    )

    print("\nLoadings:")
    display(loadings[["PC1", "PC2"]].round(3))

    plt.figure(figsize=(7, 4))
    plt.plot(
        range(1, len(cumulative) + 1),
        cumulative * 100,
        marker="o"
    )
    plt.axhline(
        90,
        linestyle="--",
        label="90% variance"
    )
    plt.xlabel("Number of principal components")
    plt.ylabel("Cumulative explained variance (%)")
    plt.title(f"{name} - PCA explained variance")
    plt.xticks(range(1, len(cumulative) + 1))
    plt.ylim(0, 105)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python

# ============================================================
# CP2 - SUBJECT-WISE PROTOCOL
# ACC-only vs ACC + cardiac
# ============================================================

import numpy as np
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

feature_sets = {
    "PAMAP2": {
        "ACC-only": [
            "acc_mean", "acc_std", "acc_min", "acc_max"
        ],
        "ACC + HR": [
            "acc_mean", "acc_std", "acc_min", "acc_max",
            "hr_mean", "hr_std"
        ]
    },
    "MHEALTH": {
        "ACC-only": [
            "acc_mean", "acc_std", "acc_min", "acc_max"
        ],
        "ACC + ECG": [
            "acc_mean", "acc_std", "acc_min", "acc_max",
            "ecg_mean", "ecg_std", "ecg_min", "ecg_max"
        ]
    }
}

matrices = {
    "PAMAP2": pamap2_X,
    "MHEALTH": mhealth_X
}

protocol_results = []
protocol_data = {}

for dataset_name, df in matrices.items():

    print(f"\n===== {dataset_name} =====")

    # One split per dataset, shared by both configurations
    splitter = GroupShuffleSplit(
        n_splits=1,
        test_size=0.2,
        random_state=42
    )

    train_idx, test_idx = next(
        splitter.split(
            df,
            groups=df["subject_id"]
        )
    )

    train_subjects = set(
        df.iloc[train_idx]["subject_id"]
    )
    test_subjects = set(
        df.iloc[test_idx]["subject_id"]
    )

    assert train_subjects.isdisjoint(test_subjects)

    print("Train subjects:", sorted(train_subjects))
    print("Test subjects:", sorted(test_subjects))

    for configuration, columns in feature_sets[
        dataset_name
    ].items():

        X = df[columns].copy()
        X = X.replace([np.inf, -np.inf], np.nan)

        if X.isna().any().any():
            raise ValueError(
                f"{dataset_name} / {configuration}: "
                "missing values detected"
            )

        X_train = X.iloc[train_idx]
        X_test = X.iloc[test_idx]

        scaler = StandardScaler()

        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)

        pca = PCA()

        train_pca = pca.fit_transform(
            X_train_scaled
        )

        test_pca = pca.transform(
            X_test_scaled
        )

        cumulative = np.cumsum(
            pca.explained_variance_ratio_
        )

        n90 = (
            np.searchsorted(cumulative, 0.90) + 1
        )

        protocol_results.append({
            "Dataset": dataset_name,
            "Configuration": configuration,
            "Train subjects": len(train_subjects),
            "Test subjects": len(test_subjects),
            "Train windows": len(train_idx),
            "Test windows": len(test_idx),
            "Features": len(columns),
            "PC1 (%)": round(
                pca.explained_variance_ratio_[0] * 100, 2
            ),
            "PC1+PC2 (%)": round(
                cumulative[1] * 100, 2
            ),
            "PCs for 90%": int(n90)
        })

        protocol_data[
            (dataset_name, configuration)
        ] = {
            "scaler": scaler,
            "pca": pca,
            "train_indices": train_idx,
            "test_indices": test_idx,
            "train_pca": train_pca,
            "test_pca": test_pca,
            "features": columns
        }

results_df = pd.DataFrame(protocol_results)

print("\n===== CP2 COMPARISON =====")
display(results_df)
```

```python

# ============================================================
# CP3 - PCA vs UMAP (subject-wise protocol)
# ============================================================

import numpy as np
import pandas as pd
import umap

from sklearn.manifold import trustworthiness
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

cp3_results = []
cp3_embeddings = {}

for (dataset_name, configuration), saved in protocol_data.items():

    print(f"\n===== {dataset_name} / {configuration} =====")

    df = matrices[dataset_name]
    columns = saved["features"]

    train_idx = saved["train_indices"]
    test_idx = saved["test_indices"]

    X_train = df.iloc[train_idx][columns]
    X_test = df.iloc[test_idx][columns]

    # Fit scaling only on training data
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # PCA: 2-dimensional representation
    pca = PCA(n_components=2)
    Z_train_pca = pca.fit_transform(X_train_scaled)
    Z_test_pca = pca.transform(X_test_scaled)

    # UMAP: 2-dimensional representation
    reducer = umap.UMAP(
        n_components=2,
        n_neighbors=15,
        min_dist=0.1,
        random_state=42
    )

    Z_train_umap = reducer.fit_transform(X_train_scaled)
    Z_test_umap = reducer.transform(X_test_scaled)

    # Trustworthiness on held-out subjects
    k = 10

    tw_pca = trustworthiness(
        X_test_scaled, Z_test_pca, n_neighbors=k
    )

    tw_umap = trustworthiness(
        X_test_scaled, Z_test_umap, n_neighbors=k
    )

    for method, tw in [
        ("PCA", tw_pca),
        ("UMAP", tw_umap)
    ]:
        cp3_results.append({
            "Dataset": dataset_name,
            "Configuration": configuration,
            "Method": method,
            "Trustworthiness": round(tw, 4),
            "Train windows": len(train_idx),
            "Test windows": len(test_idx)
        })

    cp3_embeddings[(dataset_name, configuration)] = {
        "PCA": Z_test_pca,
        "UMAP": Z_test_umap,
        "test_indices": test_idx
    }

cp3_results_df = pd.DataFrame(cp3_results)

print("\n===== CP3 - PCA vs UMAP =====")
display(cp3_results_df)
```

```python

# ============================================================
# CP3 - Stability analysis: PCA vs UMAP
# Same subject-disjoint protocol as CP2
# ============================================================

import numpy as np
import pandas as pd
import umap

from sklearn.decomposition import PCA
from sklearn.manifold import trustworthiness

seeds = [42, 123, 2026]
stability_results = []

datasets = {
    "PAMAP2": pamap2_X,
    "MHEALTH": mhealth_X
}

for (dataset_name, configuration), data in protocol_data.items():

    print(f"\n===== {dataset_name} / {configuration} =====")

    X = datasets[dataset_name]
    features = data["features"]

    train_idx = data["train_indices"]
    test_idx = data["test_indices"]

    # Reuse the scaler fitted on training data in CP2
    scaler = data["scaler"]

    X_train = X.iloc[train_idx][features].to_numpy()
    X_test = X.iloc[test_idx][features].to_numpy()

    X_train_scaled = scaler.transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # PCA baseline: fit on train, transform test
    pca = PCA(n_components=2, svd_solver="full")
    pca.fit(X_train_scaled)
    Z_pca = pca.transform(X_test_scaled)

    tw_pca = trustworthiness(
        X_test_scaled,
        Z_pca,
        n_neighbors=5
    )

    # UMAP stability across 3 random seeds
    for seed in seeds:

        reducer = umap.UMAP(
            n_components=2,
            n_neighbors=15,
            min_dist=0.1,
            random_state=seed,
            transform_seed=seed,
            n_jobs=1
        )

        reducer.fit(X_train_scaled)
        Z_umap = reducer.transform(X_test_scaled)

        tw_umap = trustworthiness(
            X_test_scaled,
            Z_umap,
            n_neighbors=5
        )

        stability_results.append({
            "Dataset": dataset_name,
            "Configuration": configuration,
            "Seed": seed,
            "PCA": tw_pca,
            "UMAP": tw_umap
        })

stability_df = pd.DataFrame(stability_results)

print("\n===== RESULTS BY SEED =====")
display(stability_df.round(4))

stability_summary = (
    stability_df
    .groupby(["Dataset", "Configuration"])
    .agg(
        PCA_mean=("PCA", "mean"),
        PCA_std=("PCA", "std"),
        UMAP_mean=("UMAP", "mean"),
        UMAP_std=("UMAP", "std")
    )
    .reset_index()
)

print("\n===== CP3 STABILITY SUMMARY =====")
display(stability_summary.round(4))

stability_summary.to_csv(
    "/kaggle/working/CP3_stability_summary.csv",
    index=False
)

print("\nSaved: CP3_stability_summary.csv")
```

```python

# ============================================================
# CP3 - Cross-modal temporal lag correlations
# ACC vs HR / ECG
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

datasets = {
    "PAMAP2": {
        "data": pamap2_X,
        "acc": "acc_std",
        "cardiac": "hr_mean"
    },
    "MHEALTH": {
        "data": mhealth_X,
        "acc": "acc_std",
        "cardiac": "ecg_std"
    }
}

lags = [-3, -2, -1, 0, 1, 2, 3]
window_seconds = 5

lag_results = []

for name, config in datasets.items():

    X = config["data"].copy()
    acc_col = config["acc"]
    cardiac_col = config["cardiac"]

    print(f"\n===== {name}: {acc_col} vs {cardiac_col} =====")

    # Keep chronological order within each subject
    X = X.sort_values(["subject_id", "window_id"])

    for lag in lags:

        paired_parts = []

        for subject, group in X.groupby("subject_id"):

            group = group.sort_values("window_id").copy()

            # Positive lag = cardiac signal occurs later
            group["cardiac_shifted"] = (
                group[cardiac_col].shift(-lag)
            )

            # Ensure windows are exactly 'lag' steps apart
            group["shifted_window_id"] = (
                group["window_id"].shift(-lag)
            )

            valid = group[
                (group["shifted_window_id"] -
                 group["window_id"] == lag)
            ][[acc_col, "cardiac_shifted"]].dropna()

            paired_parts.append(valid)

        paired = pd.concat(paired_parts, ignore_index=True)

        r = paired[acc_col].corr(
            paired["cardiac_shifted"]
        )

        lag_results.append({
            "Dataset": name,
            "ACC feature": acc_col,
            "Cardiac feature": cardiac_col,
            "Lag (seconds)": lag * window_seconds,
            "Pearson r": r,
            "Pairs": len(paired)
        })

lag_df = pd.DataFrame(lag_results)

print("\n===== CP3 LAG CORRELATIONS =====")
display(lag_df.round(4))

# Plot results
fig, ax = plt.subplots(figsize=(9, 5))

for name, group in lag_df.groupby("Dataset"):
    ax.plot(
        group["Lag (seconds)"],
        group["Pearson r"],
        marker="o",
        label=name
    )

ax.axvline(0, linestyle="--", color="gray")
ax.set_xlabel("Temporal lag (seconds)")
ax.set_ylabel("Pearson correlation (r)")
ax.set_title("CP3 - ACC vs cardiac temporal relationships")
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    "/kaggle/working/CP3_lag_correlations.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

# Save numerical results
lag_df.to_csv(
    "/kaggle/working/CP3_lag_correlations.csv",
    index=False
)

print("\nSaved: CP3_lag_correlations.png")
print("Saved: CP3_lag_correlations.csv")
```

```python

# ============================================================
# CP3 - Cross-modal relationship maps (ACC vs HR / ECG)
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

output_dir = Path("/kaggle/working")
relationship_results = []

datasets = {
    "PAMAP2": (pamap2_X, "hr_"),
    "MHEALTH": (mhealth_X, "ecg_")
}

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, (name, (df, cardiac_prefix)) in zip(
    axes, datasets.items()
):
    acc_cols = [
        c for c in df.columns
        if c.startswith("acc_")
    ]

    cardiac_cols = [
        c for c in df.columns
        if c.startswith(cardiac_prefix)
        and c != "hr_samples"
    ]

    corr = df[acc_cols + cardiac_cols].corr(
        method="pearson"
    ).loc[acc_cols, cardiac_cols]

    for acc in acc_cols:
        for cardiac in cardiac_cols:
            relationship_results.append({
                "Dataset": name,
                "ACC feature": acc,
                "Cardiac feature": cardiac,
                "Pearson r": corr.loc[acc, cardiac]
            })

    im = ax.imshow(
        corr.values,
        cmap="coolwarm",
        vmin=-1,
        vmax=1,
        aspect="auto"
    )

    ax.set_xticks(range(len(cardiac_cols)))
    ax.set_xticklabels(cardiac_cols, rotation=45, ha="right")

    ax.set_yticks(range(len(acc_cols)))
    ax.set_yticklabels(acc_cols)

    ax.set_title(f"{name}: ACC vs cardiac")

    for i in range(len(acc_cols)):
        for j in range(len(cardiac_cols)):
            value = corr.iloc[i, j]
            ax.text(
                j, i, f"{value:.2f}",
                ha="center",
                va="center",
                fontsize=10,
                color="black"
            )

fig.colorbar(
    im,
    ax=axes,
    label="Pearson correlation",
    shrink=0.85
)

fig.suptitle(
    "CP3 - Cross-modal relationships",
    fontsize=14
)

fig.savefig(
    output_dir / "CP3_cross_modal_relationships.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

relationship_df = pd.DataFrame(relationship_results)

print("===== CP3 CROSS-MODAL RELATIONSHIPS =====")
display(relationship_df.round(4))

relationship_df.to_csv(
    output_dir / "CP3_cross_modal_relationships.csv",
    index=False
)

print("Files saved successfully.")
```

```python

# ============================================================
# CP3 - Subject-level cross-modal correlations
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

subject_results = []

configs = [
    ("PAMAP2", pamap2_X, "acc_std", "hr_mean"),
    ("MHEALTH", mhealth_X, "acc_std", "ecg_std")
]

for dataset_name, df, acc_feature, cardiac_feature in configs:

    print(f"\n===== {dataset_name} =====")

    for subject, group in df.groupby("subject_id"):

        valid = group[
            [acc_feature, cardiac_feature]
        ].dropna()

        if (
            len(valid) >= 3
            and valid[acc_feature].nunique() > 1
            and valid[cardiac_feature].nunique() > 1
        ):
            r = valid[acc_feature].corr(
                valid[cardiac_feature],
                method="pearson"
            )
        else:
            r = np.nan

        subject_results.append({
            "Dataset": dataset_name,
            "Subject": str(subject),
            "ACC feature": acc_feature,
            "Cardiac feature": cardiac_feature,
            "Windows": len(valid),
            "Pearson r": r
        })

subject_corr_df = pd.DataFrame(subject_results)

display(subject_corr_df.round(4))

summary = (
    subject_corr_df
    .groupby("Dataset")["Pearson r"]
    .agg(["count", "mean", "std", "min", "max"])
    .round(4)
)

print("\n===== SUBJECT-LEVEL SUMMARY =====")
display(summary)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, (dataset_name, _, _, _) in zip(axes, configs):

    subset = subject_corr_df[
        subject_corr_df["Dataset"] == dataset_name
    ].sort_values("Subject")

    ax.bar(
        subset["Subject"],
        subset["Pearson r"]
    )

    ax.axhline(0, color="black", linewidth=0.8)
    ax.set_ylim(-1, 1)
    ax.set_title(dataset_name)
    ax.set_xlabel("Subject")
    ax.set_ylabel("Pearson correlation (r)")
    ax.grid(axis="y", alpha=0.25)

fig.suptitle(
    "CP3 - Cross-modal correlations by subject"
)

plt.tight_layout()

plt.savefig(
    "/kaggle/working/CP3_subject_correlations.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

subject_corr_df.to_csv(
    "/kaggle/working/CP3_subject_correlations.csv",
    index=False
)

summary.to_csv(
    "/kaggle/working/CP3_subject_summary.csv"
)

print("Files saved successfully.")
```

```python

# ============================================================
# CP3 - Stability across temporal subperiods
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

period_results = []

configs = [
    ("PAMAP2", pamap2_X, "acc_std", "hr_mean"),
    ("MHEALTH", mhealth_X, "acc_std", "ecg_std")
]

for dataset_name, df, acc_col, cardiac_col in configs:

    for subject, group in df.groupby("subject_id"):

        # Preserve chronological window order
        group = group.sort_values("window_id")

        midpoint = len(group) // 2

        periods = {
            "First half": group.iloc[:midpoint],
            "Second half": group.iloc[midpoint:]
        }

        for period_name, subset in periods.items():

            valid = subset[
                [acc_col, cardiac_col]
            ].dropna()

            if (
                len(valid) >= 10
                and valid[acc_col].nunique() > 1
                and valid[cardiac_col].nunique() > 1
            ):
                r = valid[acc_col].corr(
                    valid[cardiac_col]
                )
            else:
                r = np.nan

            period_results.append({
                "Dataset": dataset_name,
                "Subject": str(subject),
                "Period": period_name,
                "Windows": len(valid),
                "Pearson r": r
            })

period_df = pd.DataFrame(period_results)

print("===== CORRELATIONS BY SUBPERIOD =====")
display(period_df.round(4))

period_pivot = period_df.pivot(
    index=["Dataset", "Subject"],
    columns="Period",
    values="Pearson r"
).reset_index()

period_pivot["Absolute difference"] = (
    period_pivot["First half"]
    - period_pivot["Second half"]
).abs()

print("\n===== SUBPERIOD STABILITY =====")
display(period_pivot.round(4))

summary = (
    period_pivot
    .groupby("Dataset")["Absolute difference"]
    .agg(["count", "mean", "std", "max"])
    .round(4)
)

print("\n===== STABILITY SUMMARY =====")
display(summary)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax, (dataset_name, _, _, _) in zip(axes, configs):

    subset = period_pivot[
        period_pivot["Dataset"] == dataset_name
    ]

    x = np.arange(len(subset))
    width = 0.35

    ax.bar(
        x - width / 2,
        subset["First half"],
        width,
        label="First half"
    )

    ax.bar(
        x + width / 2,
        subset["Second half"],
        width,
        label="Second half"
    )

    ax.set_xticks(x)
    ax.set_xticklabels(subset["Subject"])
    ax.set_ylim(-1, 1)
    ax.set_xlabel("Subject")
    ax.set_ylabel("Pearson r")
    ax.set_title(dataset_name)
    ax.legend()
    ax.grid(axis="y", alpha=0.2)

fig.suptitle(
    "CP3 - Cross-modal correlation stability by subperiod"
)

plt.tight_layout()

plt.savefig(
    "/kaggle/working/CP3_subperiod_stability.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

period_pivot.to_csv(
    "/kaggle/working/CP3_subperiod_stability.csv",
    index=False
)

print("Files saved successfully.")
```

```python

# ============================================================
# CP3 - PCA vs UMAP: Stability across test subperiods
# ============================================================

import numpy as np
import pandas as pd
from sklearn.manifold import trustworthiness
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import umap

SEEDS = [42, 123, 2026]
N_NEIGHBORS = 10

datasets = {
    "PAMAP2": pamap2_X,
    "MHEALTH": mhealth_X
}

records = []

for (dataset_name, configuration), data in protocol_data.items():

    print(f"\n===== {dataset_name} / {configuration} =====")

    df = datasets[dataset_name]
    features = data["features"]

    train_idx = np.asarray(data["train_indices"], dtype=int)
    test_idx = np.asarray(data["test_indices"], dtype=int)

    # Preserve feature names and ordering
    X_train = df.iloc[train_idx][features].copy()
    X_test = df.iloc[test_idx][features].copy()

    # Standardization fitted exclusively on train
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # PCA fitted exclusively on train
    pca = PCA(n_components=2, random_state=42)
    pca.fit(X_train_scaled)
    Z_pca = pca.transform(X_test_scaled)

    # Split test observations into two chronological halves
    # within each subject, using window_id
    test_meta = df.iloc[test_idx][["subject_id", "window_id"]].copy()
    test_meta["position"] = np.arange(len(test_meta))

    half_positions = {"First half": [], "Second half": []}

    for _, group in test_meta.groupby("subject_id"):
        group = group.sort_values("window_id")
        positions = group["position"].to_numpy()
        midpoint = len(positions) // 2

        half_positions["First half"].extend(positions[:midpoint])
        half_positions["Second half"].extend(positions[midpoint:])

    # UMAP is refitted on train for each seed
    for seed in SEEDS:

        reducer = umap.UMAP(
            n_components=2,
            n_neighbors=15,
            min_dist=0.1,
            random_state=seed,
            n_jobs=1
        )

        reducer.fit(X_train_scaled)
        Z_umap = reducer.transform(X_test_scaled)

        for period, positions in half_positions.items():

            positions = np.asarray(positions, dtype=int)

            X_part = X_test_scaled[positions]
            Z_pca_part = Z_pca[positions]
            Z_umap_part = Z_umap[positions]

            if len(positions) <= N_NEIGHBORS:
                continue

            tw_pca = trustworthiness(
                X_part,
                Z_pca_part,
                n_neighbors=N_NEIGHBORS
            )

            tw_umap = trustworthiness(
                X_part,
                Z_umap_part,
                n_neighbors=N_NEIGHBORS
            )

            records.append({
                "Dataset": dataset_name,
                "Configuration": configuration,
                "Seed": seed,
                "Period": period,
                "Test windows": len(positions),
                "PCA": round(tw_pca, 4),
                "UMAP": round(tw_umap, 4)
            })

subperiod_results = pd.DataFrame(records)

print("\n===== RESULTS BY SUBPERIOD =====")
display(subperiod_results)

subperiod_summary = (
    subperiod_results
    .groupby(["Dataset", "Configuration", "Period"])
    .agg(
        PCA_mean=("PCA", "mean"),
        PCA_std=("PCA", "std"),
        UMAP_mean=("UMAP", "mean"),
        UMAP_std=("UMAP", "std")
    )
    .reset_index()
)

print("\n===== SUBPERIOD SUMMARY =====")
display(subperiod_summary.round(4))

subperiod_results.to_csv(
    "/kaggle/working/CP3_embedding_subperiod_results.csv",
    index=False
)

subperiod_summary.to_csv(
    "/kaggle/working/CP3_embedding_subperiod_summary.csv",
    index=False
)

print("\nFiles saved successfully.")
```

```python

# ============================================================
# CP3 - Final summary and validation
# ============================================================

import numpy as np
import pandas as pd
from pathlib import Path

output_dir = Path("/kaggle/working")

summary = subperiod_summary.copy()

# Compare first and second halves
comparison = summary.pivot(
    index=["Dataset", "Configuration"],
    columns="Period",
    values=["PCA_mean", "UMAP_mean"]
)

final_rows = []

for (dataset, config), row in comparison.iterrows():

    pca_first = row[("PCA_mean", "First half")]
    pca_second = row[("PCA_mean", "Second half")]

    umap_first = row[("UMAP_mean", "First half")]
    umap_second = row[("UMAP_mean", "Second half")]

    pca_average = (pca_first + pca_second) / 2
    umap_average = (umap_first + umap_second) / 2

    final_rows.append({
        "Dataset": dataset,
        "Configuration": config,
        "PCA trustworthiness": round(pca_average, 4),
        "UMAP trustworthiness": round(umap_average, 4),
        "PCA half difference": round(abs(pca_first - pca_second), 4),
        "UMAP half difference": round(abs(umap_first - umap_second), 4),
        "Higher trustworthiness": (
            "PCA" if pca_average > umap_average
            else "UMAP" if umap_average > pca_average
            else "Tie"
        )
    })

cp3_final_summary = pd.DataFrame(final_rows)

print("===== CP3 FINAL SUMMARY =====")
display(cp3_final_summary)

# Basic validation
assert len(cp3_final_summary) == 4
assert cp3_final_summary[
    ["PCA trustworthiness", "UMAP trustworthiness"]
].apply(lambda col: col.between(0, 1).all()).all()

assert not cp3_final_summary.isna().any().any()

# Export
filename = output_dir / "CP3_final_summary.csv"
cp3_final_summary.to_csv(filename, index=False)

print("\nValidation passed.")
print("Saved:", filename)
```
